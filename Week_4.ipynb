```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

x_train, x_test = x_train / 255.0, x_test / 255.0

class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']
print(f'Kích thước tập huấn luyện: {x_train.shape}')
print(f'Kích thước tập kiểm tra: {x_test.shape}')
```

```python
model = models.Sequential()

# Khối Convolutional 1
model.add(layers.Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(32, 32, 3)))
model.add(layers.BatchNormalization())
model.add(layers.Conv2D(32, (3, 3), padding='same', activation='relu'))
model.add(layers.BatchNormalization())
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Dropout(0.25)) # Ngẫu nhiên tắt 25% các kết nối ở đây

# Khối Convolutional 2
model.add(layers.Conv2D(64, (3, 3), padding='same', activation='relu'))
model.add(layers.BatchNormalization())
model.add(layers.Conv2D(64, (3, 3), padding='same', activation='relu'))
model.add(layers.BatchNormalization())
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Dropout(0.35)) # Tăng tỷ lệ dropout khi mạng sâu hơn

# Khối Convolutional 3
model.add(layers.Conv2D(128, (3, 3), padding='same', activation='relu'))
model.add(layers.BatchNormalization())
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Dropout(0.4))

# Khối Phân loại (Fully Connected)
model.add(layers.Flatten())
model.add(layers.Dense(128, activation='relu'))
model.add(layers.BatchNormalization())
model.add(layers.Dropout(0.5)) # Thường đặt dropout cao ở lớp Dense gần cuối
model.add(layers.Dense(10))    # 10 đầu ra tương ứng 10 lớp

# Xem lại cấu trúc
model.summary()
```

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# Huấn luyện mô hình với 25 epochs
history = model.fit(x_train, y_train, epochs=25,
                    batch_size=64, # Tăng batch_size để tận dụng GPU nếu có
                    validation_data=(x_test, y_test))
```

```python
import matplotlib.pyplot as plt

# Đánh giá độ chính xác trên tập test
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"\nĐộ chính xác sau khi thêm Dropout: {test_acc * 100:.2f}%")

# Vẽ biểu đồ so sánh Loss (Hàm mất mát) để thấy rõ hiện tượng Overfitting được giảm thiểu
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label = 'Val Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(loc='upper right')
plt.title('Biểu đồ Loss chống Overfitting với Dropout')
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

print("Train:", x_train.shape)
print("Test:", x_test.shape)
```

```python
class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]

plt.figure(figsize=(8,8))

for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")

plt.show()
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

print(x_train.shape)
print(x_test.shape)
```

```python
model = Sequential([

    Conv2D(
        32,
        (3,3),
        activation="relu",
        input_shape=(28,28,1)
    ),

    MaxPooling2D((2,2)),

    Conv2D(
        64,
        (3,3),
        activation="relu"
    ),

    MaxPooling2D((2,2)),

    Flatten(),

    Dense(
        128,
        activation="relu"
    ),

    Dropout(0.5),

    Dense(
        10,
        activation="softmax"
    )
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2
)
```

```python
loss, accuracy = model.evaluate(
    x_test,
    y_test
)

print("Test Accuracy:", accuracy)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    history.history["accuracy"],
    label="Train Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training History")

plt.legend()
plt.grid()

plt.show()
```

```python
index = 20

img = x_test[index]

prediction = model.predict(
    np.expand_dims(img, axis=0)
)

predicted_label = np.argmax(prediction)

plt.imshow(
    img.squeeze(),
    cmap="gray"
)

plt.title(
    f"Prediction: {class_names[predicted_label]}"
)

plt.axis("off")
plt.show()
```
